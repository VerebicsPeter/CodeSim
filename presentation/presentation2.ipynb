# Kódpárok ekvivalenciájának osztályozása

## A Sentence-BERT (SBERT) módszer

- Egy módszer a BERT modellek strukturálására
- Eredetileg NLP feladatokra használták (például mondatpárok osztályozása)
- Hatékonyan állít elő jelentéssel bíró beágyazásokat sok mondat esetén is

## Kiértékelés

- Az osztályozásra tanított hálók jobb eredményt értek el mint a metrikus tanítás
- Metrikus tanítás esetén kiértékeléshez a koszinus-hasonlóságot használtam
- $p = \frac{(sim(u,v)+1)}{2} \in {[0,1]}$
- Az osztályt $p$ alapján határoztam meg egy küszöbérték segítségével

## Kiértékelés

Eredmények a CodeNet adathalmazból mintavételezett kódpárokon vagy kódhármasokon:

<div style="font-size: 400%;">

| Model                               | Precision | Recall | F1-Score | Accuracy | Support |
|-------------------------------------|-----------|--------|----------|----------|---------|
| **BERT + Linear Classifier**        | 0.71      | 0.71   | 0.72     | 0.71     | 10,000  |
| **SBERT + Linear Classifier**       |**0.78**   |**0.78**|**0.78**  |**0.78**  | 10,000  |

A kontrasztív tanítással való finomhangolás minden küszöbérték mellett egy *biased coinflip* lett, lehetséges, hogy a *hard-negative* párok hiánya miatt.
